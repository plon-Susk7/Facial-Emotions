### Plotting resultant vectorr for IIM

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read the CSV file
df = pd.read_csv('landmark_differences_csvs/jaffe/ANG.csv')

temp_lst = []

# Compute resultant vectors
for i in range(df.shape[0]):
    row = list(df.iloc[i])
    resultant_vector = np.array([sum(row[0::2]), sum(row[1::2])])
    resultant_vector /= np.linalg.norm(resultant_vector)
    temp_lst.append(resultant_vector)

# Convert the list to a numpy array
vectors = np.array(temp_lst)

print(vectors)

# Prepare the origin points (set to (0, 0) for all vectors)
origin_x = np.zeros(vectors.shape[0])
origin_y = np.zeros(vectors.shape[0])

# Separate the x and y components of the vectors
vector_x = vectors[:, 0]
vector_y = vectors[:, 1]

# Plot the vectors
plt.figure(figsize=(8, 8))
plt.quiver(origin_x, origin_y, vector_x, vector_y, angles='xy', scale_units='xy', scale=1, color='r')
plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.grid()
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.title('2D Vectors')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.gca().set_aspect('equal', adjustable='box')
plt.show()

print(np.mean(vectors,axis=0))
```

```python
from scipy.stats import ttest_ind, ttest_rel


vector1 = [-1, -1]
vector2 = [ -1, 1]
# For independent samples
stat, p = ttest_ind(vector1, vector2)

# For paired samples
# stat, p = ttest_rel(vector1, vector2)

print(p)

if p < 0.05:
    print("The vectors are statistically different.")
else:
    print("No significant difference between the vectors.")
```

```python
from scipy.stats import mannwhitneyu

# Example:
stat, p_value = mannwhitneyu(vector1, vector2, alternative='two-sided')
if p_value < 0.05:
    print("Statistically different")
else:
    print("Not statistically different")
```
